Your five capabilities are the right five. Four changes will materially raise the hit rate.

| # | Your ask | Refinement | Why |

## 1. The narrative frame — state this before touching the keyboard

> *"It is Monday morning. You are the Executive Director of Assam Asset. Your production is running about ten percent behind target. You have fifteen workover rigs and several hundred producing wells across eight fields. Your only real lever this month is deciding where those fifteen rigs go.*
>
> *Today that decision gets made from spreadsheets, after wells have already stopped. Let me show you how it could be made before they stop."*

**Everything that follows answers one question: where do the fifteen rigs go?**

Do not describe Gemini Enterprise. Do not say "agent" in the first two minutes. Let the capability be inferred from the work.

### ACT 1 · "Where are my wells?" — 90s

> **PROMPT (type it live):**
> *"Show me all producing wells in the Geleki field on a map, sized by current oil rate and coloured by well status."*

**Expected response:** A Vega-Lite map rendered natively in the Gemini Enterprise card — field outline, well points, size by rate, colour by status (flowing / intermittent / shut-in), tooltips on hover. Accompanied by a one-line summary: *"142 wells in Geleki: 96 flowing, 19 intermittent, 27 shut-in. Current oil rate 2,840 BOPD."*

**Say while it renders:**
> *"No export. No BI request. No ticket to IT. This is your production database, answered in plain English, and the visualisation is generated on the fly — nobody built this dashboard in advance."*

**Proves:** data access · map rendering · GE-native Vega visualisation

> [!TIP]
> **Make the shut-in wells visually loud** — a distinct colour, and enough of them to be noticeable. When the room sees 27 grey dots among the live ones, someone will ask about them unprompted. That question is a gift. If nobody asks, point at them yourself and say *"hold that thought"* — it sets up Act 4.

---

### ACT 2 · "How is the asset doing?" — 60s · **DESIGNATED CUT**

> **PROMPT:**
> *"What is Geleki's oil production this month versus the same month last year, and how does that track against target?"*

**Expected response:** A Vega-Lite combination chart — monthly actual, prior year, target line — plus a one-line variance statement.

**Say:**
> *"This is the number your monthly review opens with. Notice you did not have to know which table it lives in."*

**Proves:** aggregation · Vega chart · target framing

**Cut this beat if you are more than 30 seconds behind at the end of Act 1.**

### ACT 3 · The Push-Back — 90s · **CREDIBILITY BEAT**

> **PROMPT:**
> *"Which are my lowest producing wells in Geleki?"*

**Expected response — two parts, and the second part is the point:**

*Part 1, the literal answer:* a ranked table of the five lowest-rate producers.

*Part 2, unprompted:*
> *"However — ranking by absolute rate will mislead you here. Three of these five (GK-044, GK-071, GK-098) are mature wells producing close to their expected rate for this stage of decline. There is little to recover.*
>
> *These wells are the better candidates — each is producing materially below what its own decline curve predicts:*

| Well | Actual (BOPD) | Expected from decline curve | Gap | Gap % |
|---|---|---|---|---|
| GK-112 | 14 | 38 | **−24** | −63% |
| GK-087 | 22 | 46 | **−24** | −52% |
| GK-129 | 9 | 27 | **−18** | −67% |

*Plus a Vega chart showing one well's actual-vs-fitted decline curve with the divergence shaded.*

**Say — slowly, this is the moment:**
> *"I want to draw your attention to what just happened. I asked a bad question, and it told me so. A system that only translates English into SQL would have answered the question I asked. This one answered the question I meant. That difference is the entire distance between a chatbot and something you would let near a rig schedule."*

**Proves:** domain reasoning · deterministic decline-curve fitting · willingness to disagree with the user

> [!IMPORTANT]
> **Rehearse this beat more than any other.** It is the one the room will remember and the one they will repeat to colleagues who were not there. If the push-back does not fire reliably, hard-code it — this behaviour is too important to leave to sampling.

### ACT 4 · "Who will fail, and what job do they need?" — 150s · **CORE**

> **PROMPT:**
> *"Which Geleki wells are at risk of stopping in the next 90 days, what is the likely cause in each case, and what intervention would each need?"*

**Expected response:** A ranked candidate table with visible tool execution above it.

**Show the tool calls happening.** This matters — the room must see that the arithmetic is not the language model's opinion:

```
▸ fit_decline_curve()      Arps hyperbolic, 142 wells, 36 months      ✓ 1.8s
▸ chan_diagnostic()        WOR / WOR′ log-log, water mechanism ID     ✓ 0.9s
▸ predict_stoppage_risk()  gradient-boosted classifier, 90-day window ✓ 2.1s
▸ search_well_history()    47 workover reports, 1998–2024 (GCS)       ✓ 3.4s
▸ estimate_uplift()        skin-based nodal, post-intervention rate   ✓ 1.2s
```

| Rank | Well | Stop risk, 90d | Diagnosed mechanism | Recommended intervention | Est. uplift |
|---|---|---|---|---|---|
| 1 | GK-112 | **84%** | Scale deposition in tubing | Coiled tubing acid wash | +24 BOPD |
| 2 | GK-087 | **79%** | Rod / tubing wear, deviated section | Rod string replacement + guides | +24 BOPD |
| 3 | GK-129 | **71%** | Water channelling behind casing | Cement squeeze + re-perforation | +18 BOPD |
| 4 | GK-055 | 66% | Pump wear, declining fillage | Pump changeout **(rigless — slickline)** | +11 BOPD |
| 5 | GK-103 | 58% | Bottom-water coning | Choke optimisation **(no rig required)** | +7 BOPD |

**Then drill down on ONE well — GK-129 — and only one:**

> **PROMPT:**
> *"Why do you think GK-129 is water channelling rather than coning?"*

**Expected response:** The **Chan diagnostic plot** (WOR and WOR′ against time, log-log) with the interpretation stated — *"the derivative shows a sharp early peak followed by a flattening slope, the signature of channelling through a high-permeability path, not the gradual rise characteristic of bottom-water coning"* — **plus a citation**:

> *Supporting evidence: cement bond log run 14 March 2019 recorded poor bond quality across 2,240–2,265 m. Source: `GK-129_Workover_Report_2019.pdf`, p.7.*

**Say:**
> *"Two things there. First, that distinction is not cosmetic — a cement squeeze and a water shut-off treatment cost different money and fix different problems, and treating a channelling well as a coning well wastes the rig and the barrels. Second, look at where that evidence came from. A scanned workover report from 2019. Nobody indexed that for this purpose. It found it, read it, and used it."*

**Proves:** deterministic physics (Arps, Chan) · ML risk model · unstructured document retrieval with citation · the rigless substitution insight

> [!NOTE]
> **Rows 4 and 5 are deliberately not rig jobs.** Slickline and choke optimisation remove demand from the rig queue entirely. Call this out — per the Problem Statement, rigless substitution is the one mechanism that genuinely *expands* effective capacity rather than just reordering it. An experienced well services manager in the room will notice this even if leadership does not.

### ACT 5 · "Build me the plan" — 135s

> **PROMPT:**
> *"Build next month's workover plan for Geleki and Lakwa using the 15 available rigs. Rank by deferred barrels avoided, check tool and material availability, and separate out anything that can be done rigless."*

**Expected response:**

```
ASSAM ASSET · WORKOVER PLAN · OCTOBER 2026                    [ DRAFT — AWAITING APPROVAL ]

RIG CAMPAIGN                      12 jobs · 11 rigs · 218 rig-days
  Geleki cluster    7 jobs   ← batched: single mobilisation, saves 9 rig-move days
  Lakwa cluster     5 jobs

RIGLESS CAMPAIGN                   9 jobs · slickline + CT units · 0 rig-days
                                   ← removes 9 jobs from the rig queue

MATERIALS                         ✓ 19 of 21 job kits confirmed at Nazira base
                                  ⚠ GK-129 cement squeeze — 7" packer out of stock
                                     available Sivasagar base, +2 days transit
                                     revised start 14 Oct

DEFERRED TO NOVEMBER               4 candidates · 31 BOPD at risk
                                   reason: rig capacity, lowest deferred-barrel value

─────────────────────────────────────────────────────────────────────────────
PROJECTED DEFERRED BARRELS AVOIDED       ~38,600 bbl over 12 months
EQUIVALENT                               ~106 BOPD sustained
RIG-DAYS CONSUMED                        218 of 465 available
─────────────────────────────────────────────────────────────────────────────

                    [ APPROVE ]   [ MODIFY ]   [ REJECT ]
```

**Say:**
> *"Three things I want you to notice.*
>
> *The Geleki jobs are batched — one mobilisation instead of seven, because it knew they would all be needed in the same window. That is nine rig-days recovered before a single job starts.*
>
> *It found a stock-out and rescheduled around it rather than sending a rig to wait. The CAG has costed exactly that failure mode at ONGC in the hundreds of crores.*
>
> *And four candidates were deliberately deferred, with the reason stated. It is not pretending you have capacity you do not have.*
>
> *Nothing here executes. It stops at your signature."*

**Proves:** constrained optimisation (rig capacity, materials, geography) · MRO integration · campaign batching · rigless separation · **human-in-the-loop gate**

### ACT 6 · "Change the plan" — 60s

> **PROMPT:**
> *"Add GK-114 to October — the field manager wants it done. What falls off?"*

**Expected response:**
> *"GK-114 added, priority 8, 16 rig-days. To stay within 15 rigs, LW-061 moves to November. Trade-off: you gain 9 BOPD from GK-114 and defer 14 BOPD from LW-061 — a net loss of approximately 1,800 barrels over twelve months. Recommend instead deferring GK-055, which is a rigless candidate and can run in parallel without consuming a rig."*

**Say:**

**Proves:** interactive replanning · explicit trade-off quantification · the system serving the ED rather than instructing them

## 5. What to build — technical specification

### 5.1 Data sources to stand up

Three distinct tiers, because the "data from anywhere" claim requires visibly different sources.

| Tier | Source | Contents | Why it must be separate |
|---|---|---|---|
| **Structured** | BigQuery | `well_master` (ID, field, lat/lon, completion date, lift type, TVD, deviation), `daily_production` (36 months, oil/water/gas rate, THP, choke), `well_tests`, `well_status_history` (shut-in start/end + reason code) | The map, the aggregation, the decline fits |
| **Unstructured** | GCS + Vertex AI Search | 40–50 synthetic workover reports and well completion reports as **scanned-looking PDFs**, 1998–2024 | The Act 4 citation. **This is the differentiator — do not substitute clean text files** |
| **Operational** | Cloud SQL or JSON | MRO inventory by base (Nazira, Sivasagar), rig availability calendar, job cost and duration norms | The Act 5 stock-out and the capacity constraint |

> [!IMPORTANT]
> **`well_status_history` with reason codes is the table everything depends on.** It is also precondition **P2** in the Problem Statement — the one flagged as make-or-break. Building the demo on it is deliberate: it lets you say to ONGC, *"the one thing we would need from you is this table,"* and they will already have seen why.

### 5.2 Deterministic tools the agent must call

Keep these strictly separate from the language model. The Pitch Bible position is zero LLM arithmetic, and the tool-call trace in Act 4 is how you prove it visually.

| Tool | Method | Used in |
|---|---|---|
| `fit_decline_curve()` | Arps hyperbolic, `q(t) = qᵢ / (1 + b·Dᵢ·t)^(1/b)`, least-squares | Act 3, 4 |
| `chan_diagnostic()` | WOR and WOR′ on log-log; slope classification → coning / channelling / multilayer | Act 4 |
| `predict_stoppage_risk()` | Gradient-boosted classifier, 90-day window. Features: rate trend, water-cut trend, THP drift, days since last workover, lift type, deviation, historical failure count | Act 4 |
| `estimate_uplift()` | Skin-reduction nodal estimate, IPR vs VLP | Act 4, 5 |
| `rank_candidates()` | Deferred barrels avoided over 12 months × probability of success, less job cost | Act 4, 5 |
| `schedule_rigs()` | Workover Rig Scheduling Problem. Objective = rig cost **+ deferred production** (per Aloise et al. 2006). Constraints: rig count, rig-to-job capability, geography, materials | Act 5, 6 |
| `check_mro()` | Inventory lookup with alternate-base fallback and transit penalty | Act 5 |

### 5.3 Synthetic data realism — non-negotiable parameters

An Assam Asset engineer will validate these in their head within seconds.

| Parameter | Value | Reasoning |
|---|---|---|
| Field names | **Geleki, Lakwa, Rudrasagar, Lakhmani, Laiplingaon, Demualgaon, Charali, Changmaigaon** | Real ONGC Assam Asset fields |
| Geleki on production since | 1968 | Real |
| Location | Sivasagar district, Assam — approx. 26.9°N, 94.5°E | Wells must plot in the right place. A well in the Bay of Bengal ends the demo |
| Per-well oil rate | **5–60 BOPD**, median ~20 | Derived: ONGC onshore ≈ 20–30 BOPD/well (Problem Statement §2.4). **Do not use Permian rates** |
| Geleki field total | ~2,500–3,500 BOPD | Consistent with Assam Asset ≈ 21,000 BOPD across eight fields |
| Water cut | 60–92%, rising | Mature Assam fields |
| Lift type mix | Predominantly **sucker rod pump**, some gas lift, few ESP | Per SPE-212848-PA. **Do not use the discredited 50.8% gas-lift split** |
| Rod pump MTBF | **5–9 months** | SPE-212848-PA reports 4–5 months in *problem* wells; use a slightly wider, defensible band |
| Shut-in share | ~20% of connected wells | Anchored on the Tamil Nadu census (40.5%) discounted for Assam's better geology |
| Well naming | `GK-nnn` (Geleki), `LW-nnn` (Lakwa) | Simple, readable on a map |

> [!CAUTION]
> **The most likely way this demo fails is not technical — it is a number that looks wrong to a domain expert.** A 450 BOPD onshore Assam well, a well plotted in the wrong district, or an ESP-dominated lift mix will cost you more credibility than a rendering error. Have an Assam-literate reviewer sanity-check the synthetic dataset before the rehearsal, not after.

### 5.4 The map — technical risk to retire first

Vega-Lite geographic rendering inside the Gemini Enterprise A2UI card is the **single highest technical risk in this demo.** Validate it in week one, before anything else is built.

```json
{
  "$schema": "https://vega.github.io/schema/vega-lite/v5.json",
  "width": 620, "height": 420,
  "projection": {"type": "mercator"},
  "layer": [
    {
      "data": {"values": [ /* Geleki field outline as inline GeoJSON Feature */ ]},
      "mark": {"type": "geoshape", "fill": "#0F172A", "stroke": "#00D2B4", "strokeWidth": 1.5}
    },
    {
      "data": {"values": [ /* well records, inline */ ]},
      "mark": {"type": "circle", "tooltip": true},
      "encoding": {
        "longitude": {"field": "lon", "type": "quantitative"},
        "latitude":  {"field": "lat", "type": "quantitative"},
        "size":  {"field": "oil_rate_bopd", "type": "quantitative",
                  "scale": {"range": [40, 700]}, "title": "Oil rate (BOPD)"},
        "color": {"field": "status", "type": "nominal",
                  "scale": {"domain": ["Flowing", "Intermittent", "Shut-in"],
                            "range":  ["#00D2B4", "#F59E0B", "#64748B"]}},
        "tooltip": [
          {"field": "well_id", "title": "Well"},
          {"field": "oil_rate_bopd", "title": "Oil (BOPD)"},
          {"field": "water_cut_pct", "title": "Water cut (%)"},
          {"field": "lift_type", "title": "Lift"},
          {"field": "days_since_workover", "title": "Days since workover"}
        ]
      }
    }
  ]
}
```

**Two things to confirm in week one:**
1. Does the GE renderer support `geoshape` marks and `projection`?
2. Does it permit external data URLs, or must all geometry be inline?

**Fallback ladder if `geoshape` is unsupported:**
- **Tier 1** — drop the field outline, keep `longitude`/`latitude` point encoding. Still reads as a map.
- **Tier 2** — plain `x`/`y` scatter on lon/lat with a fixed aspect ratio and a static field-outline image behind the card.
- **Tier 3** — a pre-rendered map image alongside a live Vega table. Weakest, but Act 1 is not where the demo is won.

### 5.5 Latency budget — the silent demo killer

**Any turn exceeding ~15 seconds loses the room.** Budget per act:

| Act | Target | Mitigation if over |
|---|---|---|
| 1 | < 8s | Pre-aggregate the Geleki well set into a materialised view |
| 3 | < 12s | Pre-compute all 142 decline fits nightly; the agent reads, does not fit |
| 4 | < 18s ⚠ | The heaviest beat. Pre-compute risk scores and Chan classifications. Let the tool-call trace animate so the wait feels like work, not lag |
| 5 | < 15s | Pre-solve the base schedule; solve live only for the delta |
| 6 | < 10s | Warm-start the optimiser from the Act 5 solution |

> [!TIP]
> **The visible tool-call trace in Act 4 is doing double duty.** It proves determinism *and* it makes an 18-second wait feel like the system working rather than the system hanging. Do not replace it with a spinner.

## 7. Questions the room will ask — and the answers

| Question | Who asks | Answer |
|---|---|---|
| *"Is this running on our data?"* | Anyone | No, and say so before they ask. See §6. |
| *"Our wells aren't instrumented. Half of this is manual well tests."* | Asset / Surfaces engineer | **The strongest question in the room. Concede immediately.** "Correct, and that changes what is possible. With continuous telemetry you get failure-date forecasting. With monthly well tests you get risk *ranking* — which is still enough to reorder the rig queue, and reordering the queue is where most of the value is." *(Problem Statement §6, Objection 2.)* |
| *"How accurate is the prediction?"* | Technical | Do not quote a number — there is no credible published benchmark, and vendor figures will not survive scrutiny. Say: "We would establish that against your own history in the first phase. The metric we would hold ourselves to is not prediction accuracy — it is deferred barrels avoided per rig-day." |
| *"We already have NETRA and DARPAN."* | Senior | "Yes, and they are the right foundation. Those are monitoring and visualisation. This is the planning layer on top — turning what is being watched into what gets scheduled. Your own Lakhmoni pilot with SLB showed about ten percent uplift on artificial-lift wells; this extends that logic to the rig queue." |
| *"Will it write to our systems?"* | IT / governance | "Not without a signature. Every plan stops at the approval gate you saw in Act 5. It drafts; a human authorises." |
| *"Where does the data sit?"* | Security / CISO | MeitY-compliant Indian data residency, per the standing sovereignty position. Keep it to one sentence and move on. |
| *"How long to stand this up on our data?"* | Decision maker | **The question you want.** Do not improvise. Have a rehearsed answer and the precondition list from Problem Statement §8 ready — particularly **P2, well-downtime history with reason codes.** |
| *"What if the recommended job is wrong?"* | Well services | "Then you reject it, and that rejection is training data. The system is advisory. It never touches a rig." |

## 8. Pre-flight checklist

**T-1 week**
- [ ] Vega `geoshape` rendering confirmed in the live GE app — **this gates everything**
- [ ] Synthetic dataset reviewed by someone Assam-literate
- [ ] All six prompts produce correct output on five consecutive runs
- [ ] **Act 3 push-back fires every time** — hard-code it if sampling is unreliable
- [ ] Act 4 PDF citation resolves to a real, openable document

**T-1 day**
- [ ] Full 10-minute run-through, timed, no stopping
- [ ] Second run-through with Act 2 cut, to rehearse the compressed path
- [ ] Screen-recorded backup of the complete flow, playable offline
- [ ] Static screenshots of all six outputs in a fallback deck

**T-30 minutes**
- [ ] Agent pre-warmed — run all six prompts once to fill caches
- [ ] Network verified; tethered hotspot ready
- [ ] Synthetic-data banner visible
- [ ] Browser zoom set for room legibility; notifications off
- [ ] Prompts in a text file for copy-paste — **do not risk typos on `Geleki` live**

> [!WARNING]
> **Failure protocol.** If any beat breaks, do not debug in front of the room. Say *"let me show you the result"*, cut to the screenshot, and keep the narrative moving. A demo that recovers gracefully reads as a demo of a real system. A demo that stalls while someone opens a console reads as a prototype.
