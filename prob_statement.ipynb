# Problem Statement
## Predictive Well-Stoppage Forecasting & Forward Workover Planning for ONGC Brownfield Assets

**Status:** Research draft v2.0
**Date:** 2026-09-23
**Scope:** Problem definition and evidence base only. No solution design, no architecture, no BDD.
**Decision owner in the customer organisation:** Asset Manager / Executive Director of an Asset (e.g. ED Assam Asset, ED Mumbai High, ED Mehsana).

---

## 0. BLUF — Is this a real problem?

> **Yes — but the real problem is not that wells fail. It is that well failure arrives as a surprise, and surprise converts a schedulable maintenance event into an unschedulable emergency that competes for the scarcest resource on the asset: a workover rig.**

**So what:** The value is not in knowing a well will die. It is in converting the *discovery* of a well's death from an event into a forecast, so that a fixed workover rig fleet executing **~2,200 jobs a year** can be **sequenced by deferred-barrel value** instead of by arrival order. The same rigs, better sequenced and pre-supplied, produce more barrels. No new capital, no new wells, no new reservoir.

**The four facts that make the case:**

| # | Fact | Source |
|---|---|---|
| 1 | **ONGC performs ~2,110–2,320 workover jobs and ~13,000 well stimulation jobs per year** — roughly six workovers a day, every day. This is an industrial-scale, high-frequency, repeating allocation decision. | ONGC results filings, FY2023-24 and FY2025-26 |
| 2 | **Petroleum PSU capex rose from ₹1.3 to ₹1.7 lakh crore (FY21→FY25) while domestic crude production fell from 34.2 to 28.7 MMT (FY19→FY25), a 16% decline.** COPU found MoPNG's explanation inadequate. | COPU 36th Report, August 2026 |
| 3 | **The UK regulator cut intervention cost from £11.00 to £7.60 per boe in two years — a 31% improvement — attributed to re-prioritising *what* it intervened on** (toward restoring shut-in wells). | NSTA, *2026 Wells Insights Report* |
| 4 | **ONGC has contractually conceded a +44% oil and +89% gas upside at Mumbai High from existing infrastructure**, and in June 2026 extended that premise to all 43 Western Offshore blocks. | ONGC/bp, Feb 2025 and Jun 2026 |

**Fact 2 is the thesis in one line: the binding constraint is not capital. Money went up; barrels went down.**

## 1. Who actually owns this problem

The user is correct that this does not fit a clean job-title persona, and it should not be forced into one.

The owner is the **Asset Manager / Executive Director of an Asset** — a P&L and production-target owner, not a technical discipline. In ONGC's structure this is the ED of Assam Asset, Mehsana Asset, Ahmedabad Asset, Mumbai High Asset, and so on.

| Dimension | Reality |
|---|---|
| **Primary KPI** | The asset's crude and gas production target, measured monthly, rolled up to the MoPNG MoU |
| **Controls** | A fixed workover rig fleet, a fixed well stock, an operating budget, a surveillance organisation |
| **Does not control** | Reservoir decline, crude price, the rig procurement cycle, the well stock they inherited |
| **Judged on** | Monthly, quarterly, annual — exactly the horizon on which deferred production accumulates |
| **Lacks today** | A forward, ranked, quantified view of which wells will demand a rig in the next 30/60/90 days and what each costs in barrels if it waits |

**Why this matters for framing:** a petrophysicist wants a *diagnosis*. An Asset Manager wants **a plan they can defend in a monthly production review**. The problem must therefore be stated in units of production target, rig-days and deferred barrels — not in units of pump diagnostics.

## 2. Scale of the well base — with two corrections to the public record

### 2.1 Correction #1 — "53,967 producing wells" is false

> [!CAUTION]
> **The figure "53,967 producing wells" for ONGC, which circulates in online summaries and slide decks, is wrong by roughly an order of magnitude. Do not use it.** Three independent refutations:
>
> 1. **Per-well productivity.** ONGC standalone crude FY2024-25 was 18.558 MMT ≈ **373,000 bopd**. Across 53,967 wells that is **~6.9 bopd/well** — incoherent for a company taking ~70% of its oil from offshore platforms producing 150–250 bopd per well.
> 2. **National denominator.** DGH's National Data Repository holds data for approximately **24,000 wells for all of India, all operators, since 1889.** One operator cannot hold twice every well the country has ever logged.
> 3. **ONGC's own drilling history.** ONGC has plausibly drilled **~18,000–25,000 wells in total since 1956** (derivation in §2.4). 53,967 exceeds ONGC's entire 70-year drilling history by 2–3×.

### 2.2 Correction #2 — "19–23% idle" is RIG time, not well idleness

> [!WARNING]
> The widely-quoted **19–23% figure from the CAG rig audit is rig non-productive time** — the share of *rig-days* spent not drilling. It is frequently paraphrased in a way that invites reading it as "19–23% of ONGC's wells are idle." **Those are entirely different denominators.** Additionally, the audit does not disaggregate workover-rig utilisation from drilling-rig utilisation, so the figure must not be attributed to workover rigs specifically.
>
> Also note: the rig audit is **CAG Report No. 42 of 2015**, not No. 39, which is mis-cited in much secondary coverage.

### 2.3 The one hard ONGC well census in the public domain

ONGC does not publish a consolidated producing-well count anywhere. Segment reporting goes only to Onshore vs Offshore. But one genuine, ONGC-supplied, internally-consistent census exists — **Tamil Nadu, March 2018**, disclosed under pressure during the Cauvery Delta land protests:

| Category | Count | Share |
|---|---|---|
| **Total wells drilled in Tamil Nadu** | **712** | 100% |
| Dry and abandoned | 380 | 53.4% |
| Oil/gas bearing, connected to production installations | 304 | 42.7% |
| — of which **presently flowing / operational** | **181** | 25.4% of total |
| — of which **connected but NOT flowing** | **123** | 17.3% of total |
| Utility wells | 28 | 3.9% |

*The arithmetic closes exactly: 380 + 304 + 28 = 712. Garbled or fabricated figures rarely reconcile. This reads as a genuine ONGC dataset.*

> **The operationally important ratio: 181 / 304 = 59.5% of ONGC's completed, hooked-up wells in this asset were flowing. 40.5% were connected but NOT flowing.**

**Caveat, stated plainly:** Cauvery is atypical — a 53.4% dry-hole rate reflects a geologically difficult basin, and Gujarat and Assam perform materially better. So 40.5% non-flowing-of-connected should be treated as an **upper bound**, not a point estimate.

### 2.4 Defensible inference on well population — arithmetic shown

```
(A) Onshore flowing oil producers        = 4,000 – 5,500
    [from ONGC onshore crude ≈110–120 kbopd ÷ 20–30 bopd/well,
     using Mehsana and Ahmedabad per-well rates]

(B) Flowing ÷ connected ratio            = 59.5%   [Tamil Nadu 2018]
(C) Flowing ÷ ever-drilled ratio         = 25.4%   [Tamil Nadu 2018]

Step 1  Connected onshore wells   = 4,000/0.595 to 5,500/0.595  →  ~6,700 – 9,200
Step 2  NON-FLOWING onshore       = 6,700−4,000 to 9,200−5,500  →  ~2,700 – 3,700
Step 3  Onshore drilled all-time  = 4,000/0.254 to 5,500/0.254  →  ~15,700 – 21,700
Step 4  + offshore drilled (~2,000–3,000)                       →  ~18,000 – 25,000

Cross-check: DGH NDR holds ~24,000 wells, all India, all operators, since 1889.
             ONGC ≈ 70–75% of India's drilling history → 0.72 × 24,000 ≈ 17,300.
             Two independent routes converge. Centre on ~20,000.
```

| Quantity | Estimate | Confidence |
|---|---|---|
| ONGC **cumulative wells drilled**, all-time | **~18,000–25,000** (centre ~20,000) | **Strongest** — two independent routes converge |
| ONGC **producing** oil + gas wells | ~5,000–7,500 | Moderate |
| ONGC **non-flowing connected** wells, onshore | ~2,700–3,700 (treat 3,700 as a soft ceiling; true figure likely 2,000–3,000) | **Most fragile** — generalises one atypical asset |

> [!NOTE]
> **On the working note's "more than 10,000 wells":** that is **correct at the level of cumulative well stock** and is the right framing to use. It is *not* correct for producing wells. An ONGC engineer will catch that distinction immediately.

## 4. The actual mechanism of loss

```mermaid
flowchart TD
    T0["T0 · DEGRADATION BEGINS<br/>Scale builds, pump wears, water encroaches, rods fatigue<br/><b>Signals exist in the data. Nobody is looking.</b>"]
    T1["T1 · WELL STOPS or COLLAPSES IN RATE<br/>Production loss begins accruing at full rate"]
    T2["T2 · DISCOVERY<br/>Operator round, or a zero on the daily production report<br/><i>On non-instrumented onshore wells this alone can lag hours to days</i>"]
    T3["T3 · DIAGNOSIS<br/>Pull well history, last workover, completion schematic,<br/>production trend, well tests — separate systems, some on paper"]
    T4["T4 · JOB DEFINED, RIG REQUESTED<br/>Well enters the workover queue"]
    T5["T5 · QUEUE WAIT<br/><b>Typically the longest interval.</b><br/>Competes with every other failed well, by arrival order"]
    T6["T6 · MATERIALS &amp; TOOLING MOBILISATION<br/>Possible stockout, possible inter-base transit"]
    T7["T7 · RIG MOVE + RIG UP<br/><i>CAG: 33 of 142 Assam wells saw 8–205 day mobilisation delays</i>"]
    T8["T8 · EXECUTE JOB"]
    T9["T9 · WELL RESTORED<br/>Often below pre-failure rate — near-wellbore damage<br/>accrued during shut-in, or the wrong job was done"]

    T0 --> T1 --> T2 --> T3 --> T4 --> T5 --> T6 --> T7 --> T8 --> T9

    LOSS["<b>DEFERRED BARRELS</b><br/>= integral of would-have-produced rate<br/>over the whole T1 → T9 interval"]
    T1 -.-> LOSS
    T9 -.-> LOSS
```

### 4.1 Four distinct losses, not one

| # | Loss | Mechanism |
|---|---|---|
| **1** | **Primary deferred production** | Barrels not produced between failure (T1) and restoration (T9). Irreducible once the well has failed; reducible only by shortening the interval or preventing the failure. |
| **2** | **Displacement loss (second-order)** | An emergency pulls a rig off a planned job. The planned job's barrels are deferred too. **Almost never accounted for**, because it appears as a schedule change, not a failure. |
| **3** | **Wrong-job loss** | Under time pressure with fragmented data, the intervention addresses the wrong mechanism — a scale treatment on a well that was actually watered out. Rig-days spent, barrels do not return, well re-enters the queue. |
| **4** | **Permanent recovery loss** | A well shut in long enough fails an economic revival test and is abandoned with recoverable oil behind pipe. **Deferred becomes permanent.** |

> [!NOTE]
> **A correction to the working thesis.** The original note argued continuous flow "will improve the recovery of reservoirs." True, but for a subtler reason than continuity itself. Intermittent production does cause real damage — scale and asphaltene deposition, fines migration, sand settling onto a stopped pump, water cross-flow trapping oil. But the **far more defensible recovery argument is loss #4**: wells that sit shut-in long enough get written off, and the oil behind them is never produced.
>
> The UK regulator quantifies exactly this: **558 of 2,298 UKCS wells — 24.3% of the well stock — were shut in at end-2025**, and **56 reinstated in 2025 delivered 16 mmboe** (~286 kboe per well). Frame the recovery argument as *"deferred becomes permanent"*, not *"continuous flow raises recovery factor."* The first is bulletproof; the second invites a reservoir engineer to argue.

### 4.2 Where the time actually goes

**T5, the queue wait, is the longest interval — and the only one that is purely an allocation decision rather than a physical constraint.**

T2 is an instrumentation problem. T7 and T8 are physics. T6 is supply chain. But **T5 is a choice**. Today that choice is made by arrival order and escalation pressure, because there is no forward, ranked, value-quantified queue to make it from.

### 4.3 The scale of the decision being made badly

| Metric | FY2023-24 | FY2025-26 |
|---|---|---|
| **Workover jobs completed** | **2,110** (highest ever at the time) | **2,320** (vs. plan of 2,304) |
| ONGC's own "work-over efficiency index" | 28.30 | 29.61 |
| **Well stimulation jobs** | **13,276** | **12,963** |

**Sanity check on the workover figure** *(derived)*: ONGC operates ~230 drilling + workover rigs combined. If the Assam ratio (16 drilling : 15 workover) holds company-wide, that implies ~110 workover rigs. 2,200 jobs ÷ 110 rigs ≈ **20 jobs per rig per year ≈ 18 days per job including moves** — a plausible onshore figure. The 2,200 number survives the check.

> [!IMPORTANT]
> **~2,200 workovers and ~13,000 stimulations a year is roughly six workovers and thirty-five stimulations every single day.** This is not an occasional engineering judgement. It is a high-frequency, industrial-scale, repeating allocation decision — exactly the class of decision where systematic prioritisation beats human triage, and exactly the class where a 5% sequencing improvement is worth **~110 job-equivalents of freed rig capacity per year**.
>
> Note also that **ONGC already publishes a "work-over efficiency index."** Improving a metric the customer already reports is a far easier proposition than introducing a new one. The index definition should be obtained.

## 5. The real problem, stated precisely

### Problem 1 — Detection is an event, not a forecast
Failure is discovered *after* it occurs. Degradation signals preceding most failure modes exist in the data but are not systematically surveilled across the full well population. The asset learns at T2; the correct learning point is T0.

### Problem 2 — The scarce resource is allocated without a value function
The binding constraint is **workover rig-days**, not engineering knowledge. Those rig-days are allocated by arrival order and escalation, not ranked deferred-barrel value. The peer-reviewed operations research is unambiguous here: the **Workover Rig Scheduling Problem** — canonically **Aloise, Aloise, Rocha, Ribeiro, Ribeiro Filho & Moura (2006), *Discrete Applied Mathematics* 154(5):695–702, DOI 10.1016/j.dam.2004.09.021**, validated on real Petrobras onshore instances — is **NP-complete**, and its objective function minimises rig fleet cost **plus production loss from service delay**. Optimising for rig utilisation alone is provably the wrong objective.

### Problem 3 — Under-performing wells never enter the queue at all
Reactive management is triggered by *stoppage*. A gas-lift well injecting at the wrong valve depth, a rod pump at declining efficiency, a well with rising skin — these keep producing, so they never generate a work request. **They are invisible to a stoppage-triggered system.** At a 25–33% recovery factor, this silent underperformance may exceed visible stoppage losses.

### Problem 4 — The asset has no defensible forward plan
The Asset Manager cannot answer, in a monthly production review: *"Which wells will demand a rig in the next 90 days, what will each cost me in barrels if it waits, and is my rig fleet and materials position adequate?"* Hence the last-minute scramble the working notes correctly identify.

### Problem 5 — The failure rate itself may be abnormally high
**SPE-212848-PA, *SPE Journal* Vol. 28 Iss. 3, June 2023** — "Tubing and Rod Failure Analysis in Rod Pumped Wells in an Indian Western Oil Field," Kumar, Upadhyay & Kumar — reports **MTBF as low as 4–5 months** in ONGC western onshore rod-pumped wells, against North American rod-pump benchmarks of 18–36 months. Failure mechanisms: lateral loads in deviated wells, rod and tubing buckling below the neutral point, resulting tubing holes.

> **4–5 months MTBF implies ~2.4–3.0 workovers per SRP well per year.** State it honestly: this describes *problem wells being studied*, not a fleet-wide average, so it is an upper bound on a subpopulation. Stated that way it is unimpeachable. Stated as "ONGC wells fail 4× more often" it is not.

## 6. Is it actually a problem? The honest counter-case

This section exists because the request was explicitly *"what is the real problem and if it is a problem."*

### Objection 1 — "If the rig fleet is already saturated, prediction adds nothing. You just reorder the queue."

**The strongest objection, and partly correct.** It survives only because four mechanisms create *effective* capacity:

| Mechanism | Effect |
|---|---|
| **Sequencing by value** | Same rig-days, reordered so high-deferred-barrel wells go first. More barrels at constant utilisation. |
| **Campaign batching** | Predicted failures clustered geographically, collapsing rig-move time into single mobilisations. CAG found **33 of 142 Assam wells with 8–205 day rig mobilisation delays**. |
| **Pre-positioned materials** | Lead time lets tooling arrive before the rig. CAG quantified **₹395.28 crore of idle rig cost from untimely logistics and materials**. |
| **Rigless substitution** | Early detection lets slickline / coiled tubing / through-tubing substitute for a rig workover. **This genuinely expands capacity** by removing demand from the queue. |

> **Verdict: does not defeat the thesis, but it defines the success metric.** The metric is *not* "failures predicted." It is **deferred barrels avoided per rig-day deployed.** Any framing that stops at prediction accuracy measures the wrong thing.
>
> **The regulator has already proven this mechanism works.** NSTA cut UKCS intervention cost from **£11.00/boe (2023) → £9.60 (2024) → £7.60 (2025)** — a **31% improvement in two years** — attributed to deliberately shifting focus toward restoring shut-in wells rather than other intervention types. That is a regulator, not a vendor, demonstrating that **changing what you choose to intervene on produces a measurable capital-efficiency gain.** It is the single most on-point external citation available.

### Objection 2 — "Well failure is stochastic. You cannot predict it."

**Partly true. The claim must be narrowed.**

| Predictability | Failure modes |
|---|---|
| **Good** | Rod pump wear and rod failure (dynamometer/load signatures); ESP degradation (motor amperage, intake pressure, motor temperature, drive frequency); scale build-up; sand production |
| **Moderate** | Tubing leaks; gas-lift valve failure; progressive water encroachment |
| **Poor** | Sudden mechanical failure; casing collapse; external events (power outage, monsoon access loss) |

And predictability depends entirely on **measurement frequency**. A well tested manually once a month carries almost no predictive signal.

> **The defensible claim:** *A meaningful subset of well stoppages is forecastable with useful lead time. For the remainder, wells can still be **risk-ranked** even where the date cannot be forecast. Risk ranking alone is sufficient to change rig sequencing, which is where the value is.*

Published precedent exists — **SPE-224462-MS**, "Application of AI Foresight Model in ADNOC Offshore for Preventing Premature ESP Failure and Extending Overall Run Life," Abdelkerim, Alseedi & Al-Radhi, GOTECH, 21–23 April 2025 — describing a hybrid ML + physics + digital twin approach whose stated benefit is **optimised rig allocation and inventory**, i.e. exactly the framing above.

### Objection 3 — "ONGC already has digital initiatives. This is solved."

ONGC has real programmes: **NETRA** (IPEOT, "Bytes to Barrel", real-time analytics and well/network modelling), **DARPAN** (corporate visualisation centre), **RTOC** (Real-Time Operations Centre, Dehradun, drilling-focused).

But these are predominantly **drilling-and-monitoring** oriented, and **monitoring is not forward planning**. None is publicly documented as producing a ranked, economically-quantified, forward workover queue for an Asset Manager.

**And ONGC has already proved the value internally:** the ONGC–SLB digital-oilfield pilot at **Lakhmoni field demonstrated ~10% production enhancement in artificial-lift wells.**

> **Verdict: this objection strengthens the case.** It removes "will this work here?" risk and reframes the work as extending a proven internal pilot into the planning layer.

### Objection 4 — "Is the loss actually large enough to matter?"

| Anchor | Figure | Source |
|---|---|---|
| UKCS production efficiency | **77% (2023), 75% (2024), 76% (2025)** → ~24% of potential undelivered | NSTA, *UKCS Production Efficiency in 2025*, 6 Aug 2026 |
| UKCS shut-in well stock | **558 of 2,298 wells = 24.3%** shut in at end-2025 | NSTA, *2026 Wells Insights Report* |
| UKCS reinstatement value | **56 shut-in wells reinstated in 2025 → 16 mmboe** (~286 kboe/well) | NSTA, *2026 Wells Insights Report* |
| ONGC rig management losses FY2010-14 | **₹7,995 crore total**, of which **₹6,418 crore controllable**; production deferment **≥ ₹5,117 crore** | CAG Report No. 42 of 2015 |
| ONGC Western Offshore water injection FY2014-19 | **3.79 MMT crude lost, ₹11,276 crore** | CAG of India |
| ONGC Mumbai High gas flaring FY2012-20 | **₹816.08 crore** flared, from unavailable standby compressors | CAG of India, Dec 2021 |
| Indonesia idle wells (NOC analogue) | **~16,000–17,000 idle**, **~4,495 with reactivation potential**; Pertamina holds 4,886 of which ~4,200 reactivatable; SKK Migas targeting **1,000–1,500 reactivations/year** to 2028 under Permen ESDM 14/2025 | SKK Migas / ESDM / Pertamina |

> [!IMPORTANT]
> **The CAG finding that matters most: ₹6,418 crore of ₹7,995 crore in rig-related losses was attributed to factors *controllable by ONGC*.** That is the national auditor establishing that ONGC's production gap is an **execution and allocation problem**, not a geology problem.

> [!WARNING]
> **A finding that cuts against the pitch, and which must be handled honestly before a client raises it.** NSTA attributes **83 of 114 mmboe of 2025 UKCS production losses — about 73% — to plant and facilities issues, not wells.** Well-related losses are a *minority* of the offshore loss pool.
>
> **Do not pitch "24% of production is lost and we fix it."** A technically literate client will correctly object that most of that is topsides. **Correct framing:** use the PE gap to establish that *a large, regulator-measured, routinely-quantified loss pool exists*, then pivot to well-specific numbers (shut-in stock, reinstatement economics, intervention cost/boe) for the actual addressable slice.
>
> There is a reasoned asymmetry in our favour: **UKCS is offshore, facility-heavy and ESP/gas-lift dominated. A mature onshore rod-pump field has a structurally different loss mix with a much higher well-related share** — consistent with the 4–5 month SRP MTBF in SPE-212848-PA. State that as a reasoned expectation, **not as a cited number**: no source for the onshore split was found.

### Objection 5 — "The data won't be there."

**The most serious practical risk, and not fully answerable from public sources.**

Much of ONGC's onshore well stock is understood to be on manual well testing at daily-to-monthly frequency rather than continuous telemetry. Prediction on sparse, irregular, manually-recorded data is a fundamentally different problem from prediction on high-frequency SCADA. Oil India's comparable DRIVE programme instrumented **77 production wells with 482 field devices** — indicating instrumentation in Indian onshore operations is deployed selectively, not universally.

> **Verdict: a precondition, not an objection.** See §8.

**Frame B — rig capacity released, company-wide *(the cleaner framing)*:**

```
ONGC workover jobs per year                          ≈ 2,200
A 5% improvement in sequencing / wait-time reduction ≈ 110 job-equivalents
                                                        of freed rig capacity per year
```

Frame B is preferable in a customer conversation because it is denominated in a unit ONGC already counts and already reports an efficiency index against — no assumption about deferred-barrel share is required to state it.

**The honest statement of value:**

> *We do not yet know how large the prize is at any specific ONGC asset, because no one has measured it — ONGC does not publish it and the CAG has never audited it. What we know is that (a) the loss mechanism is real and structurally present, (b) the national auditor has three times found multi-thousand-crore controllable production losses at ONGC, (c) the parliamentary committee has found capex rising 31% while national production fell 16% and rejected the explanation, (d) ONGC has contractually conceded a 44% oil upside on its flagship brownfield, (e) ONGC's own Lakhmoni pilot produced ~10% uplift on artificial-lift wells, and (f) the UK regulator has demonstrated a 31% intervention cost-per-barrel improvement purely by re-prioritising the intervention queue. The first deliverable of any engagement should be to measure the gap, not to assume it.*

## 8. What would have to be true — falsifiable preconditions

| # | Precondition | How to test | If it fails |
|---|---|---|---|
| **P1** | **Well-level production history at usable frequency** — daily rates or well tests at ≥ weekly cadence, ≥ 24 months, ≥ 80% of producers | Request a sample extract for one field | Fall back to risk-ranking on well attributes and event history rather than time-series forecasting |
| **P2** | **Well downtime and status history recorded** — start and end of every shut-in, with reason codes | Inspect the production database / daily production report structure | **Hardest blocker.** Cannot measure deferred production, cannot establish a baseline, cannot prove value |
| **P3** | **Workover job history retrievable** — job type, date, duration, well, outcome | Inspect the CMMS / workover records | Cannot learn failure signatures or estimate job durations; scheduling degrades to capacity-only |
| **P4** | **Artificial lift telemetry on a meaningful subset** — ESP amps/temp/intake pressure, or rod pump dynamometer cards | Asset instrumentation survey | Restricts predictive scope to rate-decline mechanisms (scale, watering out); excludes equipment failure prediction |
| **P5** | **The rig queue is actually a decision** — the asset has discretion over sequencing | Interview the asset's well services planner | Value collapses to diagnosis support only |
| **P6** | **Deferred production is or can be valued** | Interview the Asset Manager | The optimisation has no objective function |
| **P7** | **The "work-over efficiency index" definition is obtainable** | Ask ONGC; check investor filings | Lose the opportunity to improve a metric the customer already reports |

> [!WARNING]
> **P2 is make-or-break.** Without recorded well-downtime history with reason codes, there is no baseline, no deferred-production computation, and no way to prove anything delivered has worked. **If P2 fails, the first project is a production-loss-accounting project, not a prediction project.**

## 10. Summary — the problem in one page

**Context.** ONGC produces ~65–70% of India's crude and ~84% of its gas from a well stock it has built over 70 years — plausibly ~20,000 wells drilled, ~5,000–7,500 currently producing — concentrated in fields 43 to 68 years old, at a recovery factor of 25–33%. India's crude import dependency has risen every year since the 2015 reduction target and stands at a record ~88.7%.

**The current operating model.** Well interventions are triggered *after* a well stops or collapses in rate. Diagnosis is assembled manually from fragmented systems. The well joins a workover queue served by a fixed, scarce rig fleet — 15 rigs on the Assam Asset — sequenced largely by arrival order and escalation pressure rather than economic value. ONGC executes roughly **2,200 workovers and 13,000 stimulations a year** this way.

**The consequence.** Four compounding losses: primary deferred production between failure and restoration; displacement losses when emergencies pre-empt planned jobs; wrong-job losses from time-pressured diagnosis; and permanent recovery loss when a well sits shut-in long enough to be written off with oil behind pipe. Meanwhile, wells that are merely *underperforming* rather than stopped never enter the queue at all.

**The evidence it is real and material.** The CAG has three times found multi-thousand-crore production losses at ONGC from controllable operational execution — most pointedly **₹6,418 crore of ₹7,995 crore in rig-related losses attributed to factors within ONGC's own control**. The parliamentary COPU found **capex up 31% while national production fell 16%**, and rejected MoPNG's explanation. ONGC's Assam Asset is running **10.66% below target**. ONGC has contractually conceded, via bp, that **44% more oil** is recoverable from Mumbai High using existing infrastructure — extended to all 43 Western Offshore blocks. SPE-212848-PA reports **4–5 month rod pump MTBF** in ONGC's western onshore wells against an 18–36 month benchmark. And the UK regulator has demonstrated a **31% reduction in intervention cost per barrel purely by re-prioritising which wells it intervenes on**.

**The real problem, stated once.**

> **The Asset Manager owns a production target, and the principal lever available to defend it is the sequencing of a scarce workover rig fleet executing thousands of jobs a year — yet that fleet is dispatched reactively, by arrival order, with no forward view of which wells will fail, when, or what each will cost in barrels if it waits. The failure is not that wells die. It is that their deaths are a surprise, and surprise forfeits the only degree of freedom the Asset Manager actually has.**

**The falsifiable success metric.** Not prediction accuracy. **Deferred barrels avoided per rig-day deployed.**

## 11. Mumbai High — why the circulating well figures are stale

The widely-repeated **665 oil wells / 90 gas wells / >1,300 water injectors** figure set traces only to Scribd and SlideShare, and could not be dated against ONGC, bp or DGH material. Datable redevelopment activity shows why it cannot be current:

| Phase | Wells | Status |
|---|---|---|
| Mumbai High **North** Phase III (approved Jun 2014) | **52 new + 24 sidetracks = 76**; 5 new wellhead platforms; mods at 13 platforms | Completed 2017 |
| Mumbai High **North** Phase IV | **43 wells** | Completed Dec 2022; new platform Feb 2023 |
| Mumbai High **South** Phase III (approved 2014) | **70 new wells** (57 done by Sep 2018); 3 new WHPs, 2 clamp-ons, mods at 18 platforms | Largely complete |
| Mumbai High **South** Phase IV | Water injection facility upgrade; new NWIS-R platform | — |
| **Phase V** (field-wide, current) | **71 wells = 16 new + 55 sidetracks**; new platform RS23. Target: **4.36 MMT incremental oil + 1.83 BCM gas by 2039-40** | Underway |

**260 wells drilled or sidetracked at Mumbai High since 2014 alone.** If 665 was the pre-2014 count, the current figure should be far higher. If 665 is current, then ~260 additions were offset by a comparable number of shut-ins or conversions. **Either reading makes it unusable as a current producing-well number** — and the second reading would itself be strong evidence for this problem statement.
